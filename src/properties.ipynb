# Node neighbourhood similarity compared to node encoding distance for various structural encodings

Compare various structural encodings on how they capture the similarity of the nodes in the graphs.

```python
%reload_ext autoreload
%autoreload 2

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

PROPERTIES_PATH = "../output/properties/"
```

```python
plt.rcParams.update({'font.size': 16})
```

```python
def plot_diff_histogram(input_path, title):
    diffs = np.concatenate(pd.read_parquet(input_path)["d"].values)
    plt.hist(diffs, bins=np.unique(diffs), align="left", edgecolor="black", linewidth=1)
    plt.title(title)
    plt.xlabel("neighbourhood symmetric difference")
    plt.ylabel("number of node pairs")
    plt.show()
```

```python
def plot_sim_histogram(input_path, title):
    diffs = np.concatenate(pd.read_parquet(input_path)["sim"].values)
    plt.hist(diffs, bins=np.unique(diffs), align="left", edgecolor="black", linewidth=1)
    plt.title(title)
    plt.xlabel("neighbourhood similarity")
    plt.ylabel("number of node pairs")
    plt.show()
```

```python
def plot_path3_histogram(input_path, title):
    diffs = np.concatenate(pd.read_parquet(input_path)["path3"].values)
    plt.hist(diffs, bins=np.unique(diffs), align="left", edgecolor="black", linewidth=1)
    plt.title(title)
    plt.xlabel("number of paths of length 3 between nodes")
    plt.ylabel("number of node pairs")
    plt.show()
    plt.show()
```

```python
def plot_w_histogram(input_path, title):
    diffs = np.concatenate(pd.read_parquet(input_path)["w"].values)
    plt.hist(diffs, bins=np.unique(diffs), align="left", edgecolor="black", linewidth=1)
    plt.title(title)
    plt.xlabel("new similarity score")
    plt.ylabel("number of node pairs")
    plt.show()
    plt.show()
```

```python
def plot_diff_histogram_individual(input_path, title, i):
    diffs = np.array(pd.read_parquet(input_path)["d"].values[i])
    plt.hist(diffs, bins=np.unique(diffs), align="left", edgecolor="black", linewidth=1)
    plt.title(title + str(i))
    plt.xlabel("neighbourhood symmetric difference")
    plt.ylabel("number of node pairs")
    plt.show()
```

```python
plot_diff_histogram(PROPERTIES_PATH + "properties_ZINC.parquet", "ZINC neighbourhood distances histogram (train subset)")
```
